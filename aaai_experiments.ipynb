### Our Approah Parallelizable

```python
# Cell 1: Imports
from typing import Dict, List, Optional, Union
import pandas as pd
import numpy as np
from joblib import Parallel, delayed
from tqdm.notebook import tqdm
import os
import pickle
```

```python
# Cell 2: Initial Parameters
n_test_traces = None
window_lengths_lst = [20, None]  # Updated window lengths
n_final_markings_lst = [1]
use_heuristics = True 
unique_train_variants = False
random_seed = 101

# Define the path and file names
df_path = '/home/dsi/eli-bogdanov/Datasets/video/'
df_names = ['gtea_target_lst.pickle', 'breakfast_target_lst.pickle']

# Parameters for different datasets with correct n_train_traces
dataset_params = {
    'gtea_target_lst.pickle': 10,     # 10 traces for GTEA
    'breakfast_target_lst.pickle': 15  # 15 traces for Breakfast
}
```

```python
# Cell 3: Helper functions
def split_list_into_chunks(data_list: list, n_chunks: int) -> List[list]:
    """Split list into approximately equal chunks."""
    chunk_size = len(data_list) // n_chunks
    chunks = []
    
    for i in range(n_chunks - 1):
        start_idx = i * chunk_size
        end_idx = (i + 1) * chunk_size
        chunks.append(data_list[start_idx:end_idx])
    
    # Last chunk gets any remaining items
    chunks.append(data_list[(n_chunks-1)*chunk_size:])
    return chunks

def process_chunk(
    chunk: list,
    n_train_traces: int,
    n_test_traces: Optional[int],
    window_lengths_lst: List[Union[int, None]],
    n_final_markings_lst: List[int],
    use_heuristics: bool,
    unique_train_variants: bool,
    random_seed: int
) -> Dict:
    """Process a single chunk of tensor traces."""
    try:
        print(f"Debug: Processing chunk of size {len(chunk)}")
        
        # Convert the chunk to proper DataFrame format
        rows = []
        for trace_idx, trace in enumerate(chunk):
            case_id = f"trace_{trace_idx}"
            activities = trace.numpy().tolist() if hasattr(trace, 'numpy') else trace
            
            for activity_idx, activity in enumerate(activities):
                rows.append({
                    'case:concept:name': case_id,
                    'concept:name': str(activity),
                    'time:timestamp': pd.Timestamp.now() + pd.Timedelta(seconds=activity_idx)
                })
        
        # Create DataFrame and split into train/test
        full_df = pd.DataFrame(rows).copy()  # Create a copy to avoid warnings
        train_size = int(len(chunk) * 0.7)  # Split based on number of traces
        
        train_cases = [f"trace_{i}" for i in range(train_size)]
        test_cases = [f"trace_{i}" for i in range(train_size, len(chunk))]
        
        train_df = full_df[full_df['case:concept:name'].isin(train_cases)].copy()
        test_df = full_df[full_df['case:concept:name'].isin(test_cases)].copy()
        
        print(f"Debug: Created DataFrame with shape {full_df.shape}")
        print("Debug: DataFrame sample:")
        print(full_df.head())
        print(f"Debug: Train set size: {len(train_cases)} traces")
        print(f"Debug: Test set size: {len(test_cases)} traces")
        
        result = compare_window_based_baselines(
            df_name='',
            train_df=train_df,
            test_df=test_df,
            n_train_traces=n_train_traces,
            n_test_traces=n_test_traces,
            window_lengths_lst=window_lengths_lst,
            n_final_markings_lst=n_final_markings_lst,
            use_heuristics=use_heuristics,
            unique_train_variants=unique_train_variants,
            random_seed=random_seed
        )
        
        return {
            'status': 'success',
            'result': result,
            'chunk_size': len(chunk)
        }
    except Exception as e:
        import traceback
        return {
            'status': 'error',
            'error': str(e),
            'chunk_size': len(chunk),
            'traceback': traceback.format_exc()
        }

def process_dataset_in_chunks(
    data: list,
    df_name: str,
    n_chunks: int,
    n_train_traces: int,
    n_test_traces: Optional[int],
    window_lengths_lst: List[Union[int, None]],
    n_final_markings_lst: List[int],
    use_heuristics: bool,
    unique_train_variants: bool,
    random_seed: int,
    n_jobs: int = -1
) -> pd.DataFrame:
    """Process a dataset by splitting it into chunks."""
    print(f"\nProcessing dataset: {df_name}")
    print(f"Total traces: {len(data)}")
    print(f"Sample trace structure:")
    print(f"First trace: {data[0]}")
    
    # Split traces into chunks
    chunks = []
    chunk_size = len(data) // n_chunks
    for i in range(0, len(data), chunk_size):
        chunk = data[i:i + chunk_size]
        chunks.append(chunk)
        
    print(f"Created {len(chunks)} chunks with average size {np.mean([len(chunk) for chunk in chunks]):.0f} traces")
    
    # Process chunks in parallel
    results = Parallel(n_jobs=n_jobs)(
        delayed(process_chunk)(
            chunk=chunk,
            n_train_traces=n_train_traces,
            n_test_traces=n_test_traces,
            window_lengths_lst=window_lengths_lst,
            n_final_markings_lst=n_final_markings_lst,
            use_heuristics=use_heuristics,
            unique_train_variants=unique_train_variants,
            random_seed=random_seed
        )
        for chunk in tqdm(chunks, desc=f"Processing chunks for {df_name}")
    )
    
    # Process results
    successful_results = []
    total_processed = 0
    
    for result in results:
        if result['status'] == 'success':
            successful_results.append(result['result'])
            total_processed += result['chunk_size']
        else:
            print(f"Failed to process chunk of size {result['chunk_size']}")
            print(f"Error: {result['error']}")
            print(f"Traceback:\n{result['traceback']}")
    
    if not successful_results:
        print(f"No successful results for dataset {df_name}")
        return None
    
    print(f"Successfully processed {total_processed} traces out of {len(data)}")
    return pd.concat(successful_results, ignore_index=True)
```

```python
# Cell 4: Process datasets
for df_name in df_names:
    print(f"Processing dataset: {df_name}")
    
    # Process the pickle file
    data = process_pickle_file(os.path.join(df_path, df_name))
    
    # Adjust the number of training traces based on the dataset
    n_train_traces = dataset_params[df_name]
    
    # Determine number of chunks based on data size
    # Since the list is small, we might want to use fewer chunks
    n_chunks = max(1, min(len(data) // 5, 8))  # Adjust these numbers based on your needs
    
    # Process dataset with chunks
    result = process_dataset_in_chunks(
        data=data,
        df_name=df_name,
        n_chunks=n_chunks,
        n_train_traces=n_train_traces,
        n_test_traces=n_test_traces,
        window_lengths_lst=window_lengths_lst,
        n_final_markings_lst=n_final_markings_lst,
        use_heuristics=use_heuristics,
        unique_train_variants=unique_train_variants,
        random_seed=random_seed,
        n_jobs=60  # Adjust based on your system
    )
    
    if result is not None:
        # Define the path to save the results
        result_save_path = os.path.join(df_path, f"{df_name.split('_')[0]}_results.csv")
        
        # Save the results after processing the dataset
        save_results(result, result_save_path)
        print(f"Results saved to {result_save_path}")
```
